2. Write a Python script using requests and BeautifulSoup to scrape the names and prices of the first 20 products from the first page of a Myntra search result.

```python
import requests
from bs4 import BeautifulSoup

def scrape_myntra_products(search_term="shoes"):
    url = f"https://www.myntra.com/{search_term}"
    headers = {"User-Agent": "Mozilla/5.0"}  # helps avoid blocking
    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print("Failed to fetch page:", response.status_code)
        return

    soup = BeautifulSoup(response.text, "html.parser")

    # Find product containers
    products = soup.find_all("li", class_="product-base", limit=20)

    for idx, product in enumerate(products, start=1):
        # Extract product name
        name_tag = product.find("h3", class_="product-brand")
        title_tag = product.find("h4", class_="product-product")
        name = (name_tag.text.strip() + " " + title_tag.text.strip()) if name_tag and title_tag else "N/A"

        # Extract price
        price_tag = product.find("div", class_="product-price")
        price = price_tag.text.strip() if price_tag else "N/A"

        print(f"{idx}. {name} - {price}")

# Example usage
scrape_myntra_products("shoes")
```

3. Extend your script to loop through the first 3 pages of Zomato restaurant search results for 'pizza' in Ahmedabad, collecting restaurant names and ratings from each page.<br><br><em><strong>Hint:</strong> Pay attention to how the URL or request parameters change for each page.</em>

```python
import requests
from bs4 import BeautifulSoup

def scrape_zomato_pizza_ahmedabad(pages=3):
    base_url = "https://www.zomato.com/ahmedabad/restaurants/pizza?page={}"
    headers = {"User-Agent": "Mozilla/5.0"}  # helps avoid blocking

    all_results = []

    for page in range(1, pages + 1):
        url = base_url.format(page)
        response = requests.get(url, headers=headers)

        if response.status_code != 200:
            print(f"Failed to fetch page {page}: {response.status_code}")
            continue

        soup = BeautifulSoup(response.text, "html.parser")

        # Restaurant cards
        restaurants = soup.find_all("div", class_="sc-1mo3ldo-0")  # class may vary, inspect actual HTML

        for r in restaurants:
            # Extract restaurant name
            name_tag = r.find("h4")
            name = name_tag.text.strip() if name_tag else "N/A"

            # Extract rating
            rating_tag = r.find("div", class_="sc-1q7bklc-1")  # rating container class
            rating = rating_tag.text.strip() if rating_tag else "N/A"

            all_results.append((name, rating))

    # Print first 20 results across pages
    for idx, (name, rating) in enumerate(all_results[:20], start=1):
        print(f"{idx}. {name} - Rating: {rating}")

# Example usage
scrape_zomato_pizza_ahmedabad(pages=3)
```

4. Modify your multi-page scraping script to include a 2-second delay between each page request to avoid getting blocked by the server.<br><br><em><strong>Constraint:</strong> Use the time.sleep() function to implement rate limiting.</em>

```python
import requests
from bs4 import BeautifulSoup
import time

def scrape_zomato_pizza_ahmedabad(pages=3):
    base_url = "https://www.zomato.com/ahmedabad/restaurants/pizza?page={}"
    headers = {"User-Agent": "Mozilla/5.0"}  # helps avoid blocking

    all_results = []

    for page in range(1, pages + 1):
        url = base_url.format(page)
        response = requests.get(url, headers=headers)

        if response.status_code != 200:
            print(f"Failed to fetch page {page}: {response.status_code}")
            continue

        soup = BeautifulSoup(response.text, "html.parser")

        # Restaurant cards (class names may vary, inspect actual HTML)
        restaurants = soup.find_all("div", class_="sc-1mo3ldo-0")

        for r in restaurants:
            # Extract restaurant name
            name_tag = r.find("h4")
            name = name_tag.text.strip() if name_tag else "N/A"

            # Extract rating
            rating_tag = r.find("div", class_="sc-1q7bklc-1")
            rating = rating_tag.text.strip() if rating_tag else "N/A"

            all_results.append((name, rating))

        # Add a 2-second delay before the next request
        time.sleep(2)

    # Print first 20 results across pages
    for idx, (name, rating) in enumerate(all_results[:20], start=1):
        print(f"{idx}. {name} - Rating: {rating}")

# Example usage
scrape_zomato_pizza_ahmedabad(pages=3)
```

```python
5. Find a product listing website of your choice (not Flipkart, Myntra, or Zomato) and write a script that scrapes the first 5 product titles across multiple pages, handling pagination as needed.<br><br><em><strong>Hint:</strong> Choose a site with clear 'Next' or page number navigation.</em>
```

```python
import requests
from bs4 import BeautifulSoup

def scrape_books(pages=3):
    base_url = "http://books.toscrape.com/catalogue/page-{}.html"
    headers = {"User-Agent": "Mozilla/5.0"}

    titles = []

    for page in range(1, pages + 1):
        url = base_url.format(page)
        response = requests.get(url, headers=headers)

        if response.status_code != 200:
            print(f"Failed to fetch page {page}: {response.status_code}")
            continue

        soup = BeautifulSoup(response.text, "html.parser")

        # Each product is inside <h3> with <a> tag
        books = soup.find_all("h3")

        for book in books:
            title = book.find("a")["title"]
            titles.append(title)

            # Stop once we collect 5 titles
            if len(titles) >= 5:
                break

        if len(titles) >= 5:
            break

    # Print the collected titles
    print("First 5 book titles across pages:")
    for idx, title in enumerate(titles, start=1):
        print(f"{idx}. {title}")

# Example usage
scrape_books(pages=3)
```
