3. Compare the homepages of Myntra.com (static content) and Instagram.com (dynamic content) by reloading each site and observing the network requests in DevTools — describe one key difference in how the main content loads for each.

Myntra.com (static content)  
When you reload Myntra, most of the product listings, images, and text are delivered directly in the initial HTML response. In DevTools → Network tab, you’ll see a large HTML file and linked CSS/JS assets. The product grid is already embedded in the HTML, so the browser just renders it. This is a static content delivery model, where the server sends complete pages.

Instagram.com (dynamic content)  
Reloading Instagram shows a very different pattern. The initial HTML response is minimal — often just a shell with placeholders. The actual feed content (posts, images, captions) is fetched afterward via XHR/Fetch API calls to Instagram’s backend. In DevTools → Network tab, you’ll see multiple JSON requests after the page loads, which dynamically populate the feed. This is a dynamic content model, where JavaScript builds the page after data arrives.

3. Parse the HTML content of https://www.myntra.com using BeautifulSoup and print the title of the page.

```python
import requests
from bs4 import BeautifulSoup

# Step 1: Fetch the HTML content
url = "https://www.myntra.com"
response = requests.get(url)

# Step 2: Parse with BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Step 3: Extract and print the title
print(soup.title.string)
```

4. Extract and display all the unique class names used in <div> tags from the Zomato homepage (https://www.zomato.com) using BeautifulSoup.<br><br><em><strong>Hint:</strong> Use soup.find_all('div') and access the 'class' attribute.</em>

```python
import requests
from bs4 import BeautifulSoup

# Step 1: Fetch the Zomato homepage
url = "https://www.zomato.com"
headers = {"User-Agent": "Mozilla/5.0"}  # helps avoid blocking
response = requests.get(url, headers=headers)

# Step 2: Parse HTML with BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Step 3: Extract all <div> tags
div_tags = soup.find_all("div")

# Step 4: Collect unique class names
class_names = set()
for div in div_tags:
    if div.get("class"):
        for cls in div.get("class"):
            class_names.add(cls)

# Step 5: Display results
print("Unique class names in <div> tags:")
for cls in sorted(class_names):
    print(cls)
```
