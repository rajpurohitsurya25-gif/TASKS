2. Use Selenium to navigate to the Zomato website, wait for the search bar to load, and extract the placeholder text from the search input field.<br><br><em><strong>Hint:</strong> Use WebDriverWait with expected_conditions to wait for the element to be present before extracting its attribute.</em>

```python
# Step 1: Install Selenium and ChromeDriver
# pip install selenium
# Download ChromeDriver from https://chromedriver.chromium.org/downloads
# Ensure the driver version matches your Chrome browser.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Step 2: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 3: Open Zomato homepage
    driver.get("https://www.zomato.com")

    # Step 4: Wait for the search bar to load
    wait = WebDriverWait(driver, 10)  # wait up to 10 seconds
    search_input = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "input[placeholder]"))
    )

    # Step 5: Extract and print the placeholder text
    placeholder_text = search_input.get_attribute("placeholder")
    print("Search bar placeholder:", placeholder_text)

finally:
    # Step 6: Close the browser
    driver.quit()
```

3. Write a Selenium script to open the BookMyShow homepage, scroll down to load more content dynamically, and print the names of the first 5 movies displayed.<br><br><em><strong>Hint:</strong> Use driver.execute_script('window.scrollTo(...)') to scroll, and find_elements to get movie names.</em>

```python
# Step 1: Install Selenium and ChromeDriver
# pip install selenium
# Download ChromeDriver from https://chromedriver.chromium.org/downloads
# Ensure the driver version matches your Chrome browser.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time

# Step 2: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 3: Open BookMyShow homepage
    driver.get("https://www.bookmyshow.com")

    # Step 4: Scroll down to load more content dynamically
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)  # wait for content to load

    # Step 5: Find movie elements
    movies = driver.find_elements(By.CSS_SELECTOR, "div.__movie-name")  # inspect actual site for correct selector

    # Step 6: Print the first 5 movie names
    print("First 5 movies on BookMyShow:")
    for idx, movie in enumerate(movies[:5], start=1):
        print(f"{idx}. {movie.text}")

finally:
    # Step 7: Close the browser
    driver.quit()
```

5. Use ChatGPT or another AI tool to generate a Selenium script that logs into a demo site with a username and password, waits for a welcome message, and prints it. Then, run the script and paste the code you used along with a screenshot of the output.

```python
# Install Selenium first:
# pip install selenium
# Make sure you have ChromeDriver installed and matching your Chrome version.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Step 1: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 2: Open the demo login page
    driver.get("https://opensource-demo.orangehrmlive.com/")

    # Step 3: Enter username and password
    driver.find_element(By.ID, "txtUsername").send_keys("Admin")
    driver.find_element(By.ID, "txtPassword").send_keys("admin123")

    # Step 4: Click login button
    driver.find_element(By.ID, "btnLogin").click()

    # Step 5: Wait for welcome message to appear
    wait = WebDriverWait(driver, 10)
    welcome_msg = wait.until(
        EC.presence_of_element_located((By.ID, "welcome"))
    )

    # Step 6: Print the welcome message text
    print("Welcome message:", welcome_msg.text)

finally:
    driver.quit()
```
