1. nstall BeautifulSoup and requests in your Python environment, then write a script to fetch and print the HTML content of the Flipkart homepage.

```python
# Step 1: Install the libraries (run this in your notebook or terminal)
# !pip install requests beautifulsoup4

import requests
from bs4 import BeautifulSoup

# Step 2: Fetch the Flipkart homepage
url = "https://www.flipkart.com"
headers = {"User-Agent": "Mozilla/5.0"}  # helps avoid blocking
response = requests.get(url, headers=headers)

# Step 3: Parse the HTML with BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Step 4: Print the raw HTML content
print(soup.prettify()[:1000])  # printing first 1000 characters for readability
```

3. Write a script that uses find_all() to collect all image tags from the homepage of Myntra and prints the src attribute of each image.<br><br><em><strong>Hint:</strong> Use requests to get the HTML, then BeautifulSoup to parse and extract the src values.</em>

```python
import requests
from bs4 import BeautifulSoup

# Step 1: Fetch the Myntra homepage
url = "https://www.myntra.com"
headers = {"User-Agent": "Mozilla/5.0"}  # helps avoid blocking
response = requests.get(url, headers=headers)

# Step 2: Parse HTML with BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Step 3: Find all <img> tags
images = soup.find_all("img")

# Step 4: Print the src attribute of each image
print("Image sources found on Myntra homepage:")
for img in images:
    src = img.get("src")
    if src:  # ensure src attribute exists
        print(src)
```

4. Use BeautifulSoup's select() method with a CSS selector to extract all anchor tags inside a div with class 'trending' from a given HTML snippet (simulate a 'Trending Now' section like YouTube). Print the href attribute of each link.

```python
from bs4 import BeautifulSoup

# Sample HTML snippet simulating a 'Trending Now' section
html_doc = """
<html>
  <body>
    <div class="trending">
      <a href="https://youtube.com/video1">Trending Video 1</a>
      <a href="https://youtube.com/video2">Trending Video 2</a>
      <a href="https://youtube.com/video3">Trending Video 3</a>
    </div>
    <div class="other">
      <a href="https://youtube.com/video4">Other Video</a>
    </div>
  </body>
</html>
"""

# Step 1: Parse the HTML
soup = BeautifulSoup(html_doc, "html.parser")

# Step 2: Use CSS selector to find all anchor tags inside div.trending
links = soup.select("div.trending a")

# Step 3: Print the href attribute of each link
print("Trending links:")
for link in links:
    print(link.get("href"))
```
