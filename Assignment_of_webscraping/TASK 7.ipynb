1. Use Selenium to open https://www.myntra.com/ in a Chrome browser, locate the search box, and enter the text 'shoes' using send_keys(), then submit the search.

```python
# Install Selenium first:
# pip install selenium
# Make sure you have ChromeDriver installed and matching your Chrome version.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Step 1: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 2: Open Myntra homepage
    driver.get("https://www.myntra.com/")

    # Step 3: Wait for the search box to load
    wait = WebDriverWait(driver, 10)
    search_box = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "input.desktop-searchBar"))
    )

    # Step 4: Enter 'shoes' into the search box
    search_box.send_keys("shoes")

    # Step 5: Submit the search (press Enter)
    search_box.send_keys(Keys.RETURN)

    # Optional: wait a bit to let results load
    wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "li.product-base")))

    print("Search submitted successfully!")

finally:
    driver.quit()
```

2. Write a Selenium script that clicks the 'Login' button on https://www.zomato.com/ and handles the popup by closing it after it appears.<br><br><em><strong>Hint:</strong> Wait for the popup using WebDriverWait and locate the close button before clicking.</em>

```python
# Install Selenium first:
# pip install selenium
# Make sure you have ChromeDriver installed and matching your Chrome version.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Step 1: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 2: Open Zomato homepage
    driver.get("https://www.zomato.com/")

    # Step 3: Wait for and click the Login button
    wait = WebDriverWait(driver, 10)
    login_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//a[contains(text(),'Log in')]"))
    )
    login_button.click()

    # Step 4: Wait for the popup to appear
    popup_close_button = wait.until(
        EC.element_to_be_clickable((By.CSS_SELECTOR, "button[aria-label='Close']"))
    )

    # Step 5: Close the popup
    popup_close_button.click()

    print("Login popup handled successfully!")

finally:
    # Step 6: Close the browser
    time.sleep(2)
    driver.quit()
```

3. On https://www.flipkart.com/, use Selenium to close the initial login popup, then click on the 'Mobiles' menu item to navigate to the mobiles section.

```python
# Install Selenium first:
# pip install selenium
# Make sure you have ChromeDriver installed and matching your Chrome version.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Step 1: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 2: Open Flipkart homepage
    driver.get("https://www.flipkart.com/")

    wait = WebDriverWait(driver, 10)

    # Step 3: Close the login popup
    close_button = wait.until(
        EC.element_to_be_clickable((By.CSS_SELECTOR, "button._2KpZ6l._2doB4z"))
    )
    close_button.click()

    # Step 4: Click on the 'Mobiles' menu item
    mobiles_menu = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//div[text()='Mobiles']"))
    )
    mobiles_menu.click()

    print("Navigated to Mobiles section successfully!")

    # Optional: wait a bit to see the page
    time.sleep(3)

finally:
    driver.quit()
```

4. Build a Selenium script to scroll down the Instagram Explore page (https://www.instagram.com/explore/) multiple times, loading more posts each time, and print the total number of post links collected after each scroll.<br><br><em><strong>Hint:</strong> Use execute_script('window.scrollTo(0, document.body.scrollHeight)') and time.sleep() between scrolls.</em>

```python
# Install Selenium first:
# pip install selenium
# Make sure you have ChromeDriver installed and matching your Chrome version.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time

# Step 1: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 2: Open Instagram Explore page
    driver.get("https://www.instagram.com/explore/")

    # Optional: wait a bit for initial content to load
    time.sleep(5)

    # Step 3: Scroll multiple times
    for i in range(3):  # scroll 3 times
        # Scroll to bottom
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(3)  # wait for new posts to load

        # Collect post links
        posts = driver.find_elements(By.CSS_SELECTOR, "a[href*='/p/']")
        print(f"After scroll {i+1}, total posts collected: {len(posts)}")

finally:
    driver.quit()
```

5. Use ChatGPT or Copilot to generate a Selenium code snippet for scraping product names from an infinite scrolling page like https://www.nykaa.com/search/result/?q=lipstick. Test and modify the code so it collects at least 30 product names and prints them.

```python
# Install Selenium first:
# pip install selenium
# Make sure you have ChromeDriver installed and matching your Chrome version.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time

# Step 1: Set up ChromeDriver (update path to your chromedriver)
service = Service("path/to/chromedriver")
driver = webdriver.Chrome(service=service)

try:
    # Step 2: Open Nykaa lipstick search results
    driver.get("https://www.nykaa.com/search/result/?q=lipstick")

    # Step 3: Allow initial content to load
    time.sleep(5)

    product_names = set()

    # Step 4: Scroll until at least 30 product names are collected
    while len(product_names) < 30:
        # Scroll to bottom
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(3)  # wait for new products to load

        # Collect product name elements
        products = driver.find_elements(By.CSS_SELECTOR, "div.css-11zqbt4")  # inspect Nykaa for correct selector

        for p in products:
            name = p.text.strip()
            if name:
                product_names.add(name)

        print(f"Collected {len(product_names)} product names so far...")

    # Step 5: Print the first 30 product names
    print("\nFirst 30 product names:")
    for idx, name in enumerate(list(product_names)[:30], start=1):
        print(f"{idx}. {name}")

finally:
    driver.quit()
```
