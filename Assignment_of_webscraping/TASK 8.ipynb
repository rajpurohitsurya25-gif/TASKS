1. Use fetch() in JavaScript to get a list of movies from the public API https://api.tvmaze.com/shows and print the names of the first 10 shows in the console.

```python
 Fetch list of shows from TVMaze API
fetch("https://api.tvmaze.com/shows")
  .then(response => response.json())
  .then(data => {
    // Get the first 10 shows
    const firstTen = data.slice(0, 10);
    firstTen.forEach((show, index) => {
      console.log(`${index + 1}. ${show.name}`);
    });
  })
  .catch(error => console.error("Error fetching data:", error));
```

3. Write a function getWeather(city) that fetches current weather data for a city from the MetaWeather API (https://www.metaweather.com/api/) and prints the temperature. Use 'Ahmedabad' as the test city.<br><br><em><strong>Hint:</strong> You may need to first get the WOEID for the city, then fetch weather using that ID.</em>

```python
import requests

def getWeather(city):
    # Step 1: Get WOEID for the city
    search_url = f"https://www.metaweather.com/api/location/search/?query={city}"
    search_response = requests.get(search_url)
    search_data = search_response.json()

    if not search_data:
        print(f"No location found for {city}")
        return

    woeid = search_data[0]["woeid"]

    # Step 2: Fetch weather using WOEID
    weather_url = f"https://www.metaweather.com/api/location/{woeid}/"
    weather_response = requests.get(weather_url)
    weather_data = weather_response.json()

    # Step 3: Extract current temperature (from consolidated_weather[0])
    current_weather = weather_data["consolidated_weather"][0]
    temperature = current_weather["the_temp"]

    print(f"Current temperature in {city}: {temperature:.2f}°C")

# Test with Ahmedabad
getWeather("Ahmedabad")
```
